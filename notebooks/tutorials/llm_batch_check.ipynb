```python
from discovery_utils.utils.llm import batch_check

import pandas as pd
import os
```

```python
system_message = "Determine whether this text is about heat pumps"

fields = [
    {"name": "is_relevant", "type": "str", "description": "A one-word answer: 'yes' or 'no'."},
    {"name": "technology", "type": "str", "description": "What technology is mentioned in the text?"},
]

test_data = {
    "id2": "One type of low carbon heating is a heat pump",
    "id1": "Heat pumps are efficient",
    "id3": "Hydrogen boilers is another type of heating technology",
}

processor = batch_check.LLMProcessor(
    output_path="output.jsonl",
    system_message=system_message,
    session_name="testing",
    output_fields=fields,
)

processor.run(test_data, batch_size=1, sleep_time=0.5)
```

```python
pd.read_json("output.jsonl", lines=True)
```

```python
os.remove("output.jsonl")
```

## Example of using a config file

```python
from importlib.resources import files
path_to_config = files("discovery_utils.utils.llm").joinpath("example_relevance_check_config.yaml")

system_message = batch_check.generate_relevance_check_system_message(str(path_to_config))
```

```python
print(system_message)
```

```python
test_data = {
    "id2": "Online activity monitoring app for parents",
    "id1": "A online safety course for teenagers",
    "id3": "App for selling second-hand clothes",
}

fields = [
    {"name": "is_relevant", "type": "str", "description": "A one-word answer: 'yes' or 'no'."},
    {"name": "technology", "type": "str", "description": "What approach to online safety is mentioned in the text?"},
]

processor = batch_check.LLMProcessor(
    output_path="output.jsonl",
    system_message=system_message,
    session_name="testing",
    output_fields=fields,
)

processor.run(test_data, batch_size=3, sleep_time=0.5)
```
